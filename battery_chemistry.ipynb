```python
from os import sep
import pandas as pd
import plotly.express as px
from streamlit import header

df = pd.read_csv('sustainability_platform/data/share-of-battery-capacity-of-electric-vehicle-sales-by-chemistry-and-region-2021-2023.csv', 
                 sep=';', 
                 skiprows=3)  # Skip metadata rows
# Clean column names
df.columns = ['Year', 'Low-nickel', 'High-nickel', 'LFP']

# Melt data for Plotly
melted_df = df.melt(id_vars='Year', 
                    var_name='Chemistry', 
                    value_name='Market Share')

# Create visualization
fig = px.area(melted_df, 
              x='Year', 
              y='Market Share', 
              color='Chemistry',
              title="EV Battery Chemistry Market Share (2021-2023)")
fig.update_layout(yaxis_title="Market Share (%)")
fig.show()
```
