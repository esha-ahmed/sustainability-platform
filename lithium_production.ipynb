```python
import pandas as pd
import plotly.express as px

# Load data
df = pd.read_csv('sustainability_platform/data/lithium-production.csv')

# 1. Global Production Trend
global_df = df[df['Entity'] == 'World']
fig = px.line(global_df, x='Year', y='Lithium production - kt',
              title='Global Lithium Production (1995-2023)')
fig.show()

# 2. Top Producers 2023
latest = df[df['Year'] == 2023].sort_values('Lithium production - kt', ascending=False)[:10]
px.bar(latest, x='Entity', y='Lithium production - kt', 
       title='Top Lithium Producers (2023)')

# 3. Production Growth Analysis
countries = ['Australia', 'Chile', 'China', 'Argentina']
country_df = df[df['Entity'].isin(countries)]
px.line(country_df, x='Year', y='Lithium production - kt',
       color='Entity', title='Major Producers Timeline')
```
